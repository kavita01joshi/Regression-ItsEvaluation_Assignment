Q1. What is Simple Linear Regression?

A. Simple Linear Regression is a way to predict one thing based on another. It draws a straight line through a bunch of dots (data points) to show how two things are connected.

**Example**-

Imagine you want to know how your study time affects your exam score. You collect data from your friends:

Friend A studied 2 hours → scored 60

Friend B studied 4 hours → scored 70

Friend C studied 6 hours → scored 80

You can draw a line through these points to say: "If someone studies 5 hours, they’ll probably score around 75."

**Formula-**
The line follows this formula:

Predicted Score
=
Starting Point
+
(
Study Time
×
Slope
)
Starting Point: Where the line begins (when study time is 0)

Slope: How much the score goes up for each extra hour of study.

**Why It’s Useful-**
Helps you predict future outcomes

Shows how strongly two things are related


Q2.  What are the key assumptions of Simple Linear Regression?

**- Straight-Line Relationship -
The data should follow a straight path — like a ruler laid across the dots. If the dots curve or zigzag, this method won’t work well.

**- Each Point Is Independent -
Every data point should be its own story. One person’s score shouldn’t affect another’s.

**- Equal Spread of Mistakes -
The “mistakes” (differences between actual and predicted values) should be evenly spread out. No bunching up or wild swings.

**- Mistakes Should Be Normally Distributed -
The mistakes should follow a bell-shaped curve — most small, a few big, but not too many extremes.

**Why These Matter-**
If these rules aren’t followed, your predictions might be off or misleading — like using a ruler to measure a curvy road.


Q3.Heteroscedasticity means that the errors in your predictions are uneven.

A. Imagine you're guessing people's heights based on their ages. If your guesses are close for young kids but way off for teenagers, your errors are not consistent — that’s heteroscedasticity.

**Why It Matters-**
It makes your model less trustworthy.

It can mess up the math behind your predictions.

You might think something is important when it’s not — or miss something that is.

**What You Can Do-**
Use a log scale or other transformations.

Try different types of regression that handle uneven errors better.

Look at a plot of your errors — if they fan out like a trumpet, that’s a clue!

Q6. Implement a Python program to fit a Simple Linear Regression model to
the following sample data:
● X = [1, 2, 3, 4, 5]
● Y = [2.1, 4.3, 6.1, 7.9, 10.2]
Plot the regression line over the data points.

A.Python program using scikit-learn and matplotlib to fit a Simple Linear Regression model and plot the regression line over your data points:



```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.1, 4.3, 6.1, 7.9, 10.2])

# Create and fit the model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Plotting
plt.scatter(X, Y, color='blue', label='Actual data')
plt.plot(X, Y_pred, color='red', label='Regression line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Simple Linear Regression')
plt.legend()
plt.grid(True)
plt.show()
```

**What This Does:**
Fits a straight line to your data.

Predicts Y values using the model.

Plots both the actual data points and the regression line.

Q7. Fit a Multiple Linear Regression model on this sample data:
● Area = [1200, 1500, 1800, 2000]
● Rooms = [2, 3, 3, 4]
● Price = [250000, 300000, 320000, 370000]
Check for multicollinearity using VIF and report the results.

A. Python program that fits a Multiple Linear Regression model using scikit-learn, and checks for multicollinearity using Variance Inflation Factor (VIF) from statsmodels. Multicollinearity means that two or more predictors are too closely related, which can mess up the model’s reliability.



```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from statsmodels.stats.outliers_influence import variance_inflation_factor

# Sample data
data = {
    'Area': [1200, 1500, 1800, 2000],
    'Rooms': [2, 3, 3, 4],
    'Price': [250000, 300000, 320000, 370000]
}

df = pd.DataFrame(data)

# Features and target
X = df[['Area', 'Rooms']]
y = df['Price']

# Fit the model
model = LinearRegression()
model.fit(X, y)

# Calculate VIF
X_with_const = pd.concat([pd.Series([1]*len(X), name='Intercept'), X], axis=1)
vif_data = pd.DataFrame()
vif_data['Feature'] = X_with_const.columns
vif_data['VIF'] = [variance_inflation_factor(X_with_const.values, i) for i in range(X_with_const.shape[1])]

print("VIF Results:")
print(vif_data)
```

**What You’ll See**
The output will show VIF values for:

Intercept
Area
Rooms

If VIF is:

1–5: Usually fine

>5: Possible multicollinearity

>10: Serious multicollinearity

Q8. Implement polynomial regression on the following data:
● X = [1, 2, 3, 4, 5]
● Y = [2.2, 4.8, 7.5, 11.2, 14.7]
Fit a 2nd-degree polynomial and plot the resulting curve.

A. Python program that fits a 2nd-degree Polynomial Regression model to your data and plots the resulting curve using scikit-learn and matplotlib:



```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
Y = np.array([2.2, 4.8, 7.5, 11.2, 14.7])

# Transform to polynomial features (degree 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Fit the model
model = LinearRegression()
model.fit(X_poly, Y)

# Predict values
Y_pred = model.predict(X_poly)

# Plotting
plt.scatter(X, Y, color='blue', label='Actual data')
plt.plot(X, Y_pred, color='red', label='Polynomial regression curve')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('2nd-Degree Polynomial Regression')
plt.legend()
plt.grid(True)
plt.show()
```

**What This Does:**
Converts your input into polynomial features (adds 𝑥2).

Fits a curve to the data.

Plots both the original points and the smooth curve.

Q9.  Create a residuals plot for a regression model trained on this data:
● X = [10, 20, 30, 40, 50]
● Y = [15, 35, 40, 50, 65]
Assess heteroscedasticity by examining the spread of residuals.

A. Python program that fits a Simple Linear Regression model to your data, creates a residuals plot, and helps you visually assess heteroscedasticity:


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
X = np.array([10, 20, 30, 40, 50]).reshape(-1, 1)
Y = np.array([15, 35, 40, 50, 65])

# Fit the model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Calculate residuals
residuals = Y - Y_pred

# Plot residuals
plt.scatter(X, residuals, color='purple')
plt.axhline(y=0, color='gray', linestyle='--')
plt.xlabel('X')
plt.ylabel('Residuals')
plt.title('Residuals Plot')
plt.grid(True)
plt.show()
```

**How to Assess Heteroscedasticity-**
Look at the spread of residuals: If the dots are evenly scattered around the horizontal line (y = 0), that’s good — it suggests homoscedasticity.

If the dots fan out, cluster, or show a pattern, that could mean heteroscedasticity — uneven error spread, which can affect model reliability.


Q10. Imagine you are a data scientist working for a real estate company. You
need to predict house prices using features like area, number of rooms, and location.
However, you detect heteroscedasticity and multicollinearity in your regression
model. Explain the steps you would take to address these issues and ensure a robust
model.

A. Step-by-Step Strategy

1. **Addressing Heteroscedasticity (Uneven Error Spread)**
Heteroscedasticity means the prediction errors vary across the data — which can distort model accuracy and confidence intervals.

**Fixes:**

Log or power transformation of the target variable (e.g., use log(price) instead of price) to stabilize variance.

Use Weighted Least Squares (WLS) where observations with lower variance get more weight.

Consider robust regression techniques that adjust standard errors to handle heteroscedasticity.

**Diagnostic Tip:** Plot residuals vs. predicted values. If the spread fans out or shrinks, it’s a sign of heteroscedasticity.

2. **Addressing Multicollinearity (Highly Correlated Predictors)**
Multicollinearity makes it hard to tell which feature is really influencing the target — it inflates standard errors and weakens model interpretability.

**Fixes:**

Check VIF (Variance Inflation Factor): Drop or combine features with high VIF (typically > 5 or 10).

Use Principal Component Analysis (PCA) to reduce dimensionality while preserving variance.

Apply regularization techniques like:

Ridge Regression (L2 penalty): Shrinks coefficients to reduce impact of multicollinearity.

Lasso Regression (L1 penalty): Can eliminate irrelevant features entirely.

**Diagnostic Tip:** Use a correlation matrix or VIF scores to spot highly correlated features.

3. **Model Refinement and Validation**
Feature Engineering: Create new features (e.g., price per square foot) or encode location using one-hot or target encoding.

Cross-Validation: Use k-fold cross-validation to ensure model generalizes well.

Residual Analysis: Recheck residuals after adjustments to confirm improvements.

Model Comparison: Try different models (e.g., decision trees, random forests) to see if they handle issues better.

**Final Goal**
Build a model that:

Makes accurate predictions

Has stable and interpretable coefficients

Passes diagnostic checks for residuals and multicollinearity
